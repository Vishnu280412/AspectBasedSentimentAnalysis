```python
#Function to extract numbers from a string
import re

def extract_numbers(string):
    num_with_coma = re.search(r'\d{1,3}(?:,\d{3})*(?:\.\d+)?', string)
    return int(num_with_coma.group().replace(",", ""))
```

```python
#Clean the raw data
import pandas as pd

raw_data = ["BagaBeach", "CalanguteBeach", "RadhanagarBeach", "KalapatharBeach", "MarinaBeach", "JuhuBeach", "VarkalaBeach"]

for i in raw_data:
    # Load the CSV file into a DataFrame
    df = pd.read_csv(f'{i}.csv')

    # Filter rows where the first column is not empty
    filtered_df = df[df['BMQDV href'].notna()]

    #Renaming the columns that we want
    columnName_map = {"BMQDV":"Username", "biGQs": "Contr1", "biGQs 2": "Likes", "yCeTE": "Title", "RpeCd": "VisitedWith", "yCeTE 2": "Comment", "biGQs 4": "WrittenOn", "IugUm": "Contr2"}
    filtered_df.rename(columns=columnName_map, inplace=True)

    #Deleting unwanted columns
    columns_to_delete = ["BMQDV href", "NhWcC src", "BMQDV href 3", "biGQs 3", "biGQs 5", "lQeXP", "lQeXP href"]
    filtered_df.drop(columns=columns_to_delete, inplace=True)

    #Adding Contro Column
    filtered_df['Contr1'] = filtered_df['Contr1'].fillna('')
    filtered_df['Contr2'] = filtered_df['Contr2'].fillna('')
    filtered_df["Contribution"] = (filtered_df["Contr1"]+filtered_df["Contr2"]).apply(extract_numbers)

    #Delete two unwanted columns
    filtered_df.drop(columns=["Contr1", "Contr2"], inplace=True)

    df_no_duplicates = filtered_df.drop_duplicates(subset='Comment', keep='first')
    
    # Save the filtered DataFrame back to a CSV file
    df_no_duplicates.to_csv(f'FirstCleaning/{i}.csv', index=False)
```

```python
import pandas as pd
import nltk
from nltk.stem import PorterStemmer
from collections import Counter

df = word_counts_df

# Initialize the Porter Stemmer
stemmer = PorterStemmer()

# Function to stem and count words
def stem_and_count(text):
    stemmed_words = [stemmer.stem(word) for word in nltk.word_tokenize(text)]
    return stemmed_words

# Apply stemming to the 'words' column and create a new 'stemmed_words' column
df['stemmed_words'] = df['Word'].apply(stem_and_count)

# Initialize a Counter for the combined frequencies
combined_frequencies = Counter()

# Iterate through the DataFrame and sum the frequencies of similar words
for index, row in df.iterrows():
    for word in row['stemmed_words']:
        combined_frequencies[word] += row['Frequency']

# Convert the combined frequencies to a DataFrame
combined_df = pd.DataFrame(list(combined_frequencies.items()), columns=['Word', 'Frequency'])

# Sort the DataFrame by frequency in descending order
combined_df = combined_df.sort_values(by='Frequency', ascending=False)

# Print the combined word frequencies
print(combined_df)
combined_df.to_csv('FirstCleaning/ABSAoutput/AspectFreq/CalanguteBetterFreq.csv', index=False)
```

```python
import json
from collections import Counter
from nltk.stem import PorterStemmer

# Initialize the Porter Stemmer
stemmer = PorterStemmer()

with open('Aspect Term Extraction and Polarity Classification.FAST_LCF_ATEPC.result.json', 'r', encoding="utf8") as json_file:
    data = json.load(json_file)

# Initialize a Counter for the combined frequencies
pos_combined_frequencies = Counter()
neg_combined_frequencies = Counter()
neu_combined_frequencies = Counter()

for i in data:
    for j in range(len(i['sentiment'])):
        if i['sentiment'][j]=='Positive':
            pos_words = [stemmer.stem(word) for word in i['aspect'][j].split()]
        
        # Iterate through the stemmed words and update the frequency counter
            for word in pos_words:
                pos_combined_frequencies[word] += 1
                
        if i['sentiment'][j]=='Negative':
            neg_words = [stemmer.stem(word) for word in i['aspect'][j].split()]
        
        # Iterate through the stemmed words and update the frequency counter
            for word in neg_words:
                neg_combined_frequencies[word] += 1
        
        if i['sentiment'][j]=='Neutral':
            neu_words = [stemmer.stem(word) for word in i['aspect'][j].split()]
        
        # Iterate through the stemmed words and update the frequency counter
            for word in neu_words:
                neu_combined_frequencies[word] += 1

pos_combined_df = pd.DataFrame(list(pos_combined_frequencies.items()), columns=['Word', 'Frequency'])
pos_combined_df['Polarity'] = 'Positive'
pos_combined_df = pos_combined_df.sort_values(by='Frequency', ascending=False)

neg_combined_df = pd.DataFrame(list(neg_combined_frequencies.items()), columns=['Word', 'Frequency'])
neg_combined_df['Polarity'] = 'Negative'
neg_combined_df = neg_combined_df.sort_values(by='Frequency', ascending=False)

neu_combined_df = pd.DataFrame(list(neu_combined_frequencies.items()), columns=['Word', 'Frequency'])
neu_combined_df['Polarity'] = 'Neutral'
neu_combined_df =neu_combined_df.sort_values(by='Frequency', ascending=False)

# Print the combined word frequencies
print(pos_combined_df)
pos_combined_df.to_csv('FirstCleaning/ABSAoutput/AspectFreq/PositiveFreq/Calangute.csv', index=False)
print(neg_combined_df)
neg_combined_df.to_csv('FirstCleaning/ABSAoutput/AspectFreq/NegativeFreq/Calangute.csv', index=False)
print(neu_combined_df)
neu_combined_df.to_csv('FirstCleaning/ABSAoutput/AspectFreq/NeutralFreq/Calangute.csv', index=False)
```

```python
#Code to make data for graph of Aspect Sentiment Over Time
import ast
import pandas as pd

def remove_written(string):
    return string[9:].strip()

def remove_zero(string):
    return string[1:].strip()

#Stemming
stemmer = PorterStemmer()

merged_df2 = pd.read_csv(f'FirstCleaning/ABSAoutput/MergedOutput/Baga.csv')
new_df = pd.DataFrame(columns=['aspect', 'polarity', 'date'])

for index, row in merged_df2.iterrows():
    aspects = ast.literal_eval(row['aspect'])
    sentiments = ast.literal_eval(row['sentiment'])
    written_on = remove_written(row['WrittenOn'])
    
    try:
        written_on = pd.to_datetime(written_on, format='%d %B %Y')
    except ValueError:
        try:
            # If parsing fails, try a different format without the day component
            written_on = pd.to_datetime(written_on, format='%B %Y')
        except ValueError:
            # Handle any other parsing errors or fallback to a default date
            written_on = pd.to_datetime(remove_zero(written_on), format='%B %Y')

    
    # Iterate through the aspects and sentiments lists
    for aspect, sentiment in zip(aspects, sentiments):
        # Append new rows to new_df
        new_df = new_df.append({'aspect': aspect, 'polarity': sentiment, 'date': written_on}, ignore_index=True)

new_df.to_csv('FirstCleaning/ABSAoutput/Output_TimeSeries/Baga.csv', index=False)
```

```python
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt

#Convert the json file to csv
import pandas as pd
import json

# Load the JSON data from a file
with open('Aspect Term Extraction and Polarity Classification.FAST_LCF_ATEPC.result.json', 'r', encoding="utf8") as json_file:
    json_data = json.load(json_file)

# Create a DataFrame from the JSON data
# df = pd.DataFrame(json_data)
df=pd.read_json('FirstCleaning\ABSAoutput\MergedOutput\Calangute.json',lines=True)

# Extract aspect names
aspects = df['aspect'].explode().unique()

# Create a co-occurrence matrix
cooccurrence_matrix = pd.DataFrame(0, index=aspects, columns=aspects)

# Fill the co-occurrence matrix
for _, row in df.iterrows():
    aspect_list = row['aspect']
    for aspect1 in aspect_list:
        for aspect2 in aspect_list:
            cooccurrence_matrix.loc[aspect1, aspect2] += 1

top_aspects = cooccurrence_matrix.sum(axis=1).nlargest(10).index
# Create a network graph
G = nx.Graph()

# Add nodes and edges to the graph
for aspect in aspects:
    G.add_node(aspect)

for i, row in cooccurrence_matrix.loc[top_aspects, top_aspects].iterrows():
    for j, value in row.iteritems():
        if i != j and value > 0:
            G_top.add_edge(i, j, weight=value)

# Draw the top network graph
plt.figure(figsize=(12, 8)) 
pos_top = nx.spring_layout(G_top)
edge_labels_top = {(i, j): f"{value}" for i, j, value in G_top.edges(data='weight')}
nx.draw(G_top, pos_top, with_labels=True, font_size=14, node_size=1000, font_color='black')
nx.draw_networkx_edge_labels(G_top, pos_top, edge_labels=edge_labels_top)

plt.title("Top 10 Aspects Co-Occurrence Network")
plt.show()
```

```python
# Set a threshold for co-occurrence frequency
threshold = 5

# Filter co-occurrence matrix based on the threshold
filtered_cooccurrence_matrix = cooccurrence_matrix.applymap(lambda x: x if x > threshold else 0)

# Select the top 10 aspects based on the sum of co-occurrence frequencies
top_aspects = filtered_cooccurrence_matrix.sum(axis=1).nlargest(10).index

# Create a network graph with only the top 10 aspects
G_top = nx.Graph()

# Add nodes and edges to the top graph
for aspect in top_aspects:
    G_top.add_node(aspect)

for i, row in filtered_cooccurrence_matrix.loc[top_aspects, top_aspects].iterrows():
    for j, value in row.iteritems():
        if i != j and value > 0:
            G_top.add_edge(i, j, weight=value)

# Draw the top network graph
plt.figure(figsize=(12, 8)) 
pos_top = nx.spring_layout(G_top)
edge_labels_top = {(i, j): f"{value}" for i, j, value in G_top.edges(data='weight')}
nx.draw(G_top, pos_top, with_labels=True, font_size=14, node_size=1000, font_color='black')
nx.draw_networkx_edge_labels(G_top, pos_top, edge_labels=edge_labels_top)

plt.title("Top 10 Aspects Co-Occurrence Network")
plt.show()

```
